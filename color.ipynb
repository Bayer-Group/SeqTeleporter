```python
import numpy as np
import matplotlib.pyplot as plt

# Sample emission spectrum data (wavelengths and corresponding intensities)
wavelengths = np.array([400, 450, 500, 550, 600, 650, 700])  # Example wavelengths in nm
intensities = np.array([0, 0.5, 1, 0.8, 0.2, 0, 0])  # Example intensity values

# Normalize intensities
norm_intensities = intensities / np.max(intensities)

# Create a color based on the emission spectrum
color = plt.cm.viridis(norm_intensities)  # You can choose other colormaps

# Display the color
plt.imshow([color], aspect='auto')
plt.axis('off')
plt.show()
```
